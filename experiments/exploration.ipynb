```python
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import root_mean_squared_log_error

y = df["hours_spent"]
y_log = np.log1p(y)

def cv_score(X, use_categories=False):
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    oof = np.zeros(len(X))                 
    for train_idx, val_idx in skf.split(X, df["hours_bucket"]):
        kw = {"categorical_features": "from_dtype"} if use_categories else {}
        model = HistGradientBoostingRegressor(**kw)
        model.fit(X.iloc[train_idx], y_log.iloc[train_idx]) 
        oof[val_idx] = model.predict(X.iloc[val_idx])          
    return root_mean_squared_log_error(y, np.clip(np.expm1(oof), 0, None))



def test_skf():
    X = df.drop(columns=["project_id", "hours_spent", "log_hours", "hours_bucket"]).copy()
    for col in X.select_dtypes(include=["object", "string"]).columns:
        X[col] = X[col].astype("category")


    print(cv_score(X, use_categories=True))

test_skf()
```

Our base RMSLE is 0.674, let's try to reduce it via engineering some features:

```python
df["commits_per_active_day"] = df["num_commits"] / df["active_commit_days"]
test_skf()
```

0.668, got better

```python
df["lines_per_active_day"] = df["lines_added_total"] / df["active_commit_days"]
test_skf()
```
