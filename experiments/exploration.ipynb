```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(14, 6))

plt.scatter(df['num_commits'], df['hours_spent'], alpha=0.5, edgecolors='none')

plt.title('Approved Hours vs. Number of Commits', fontsize=14)
plt.xlabel('Number of Commits', fontsize=12)
plt.ylabel('Approved Hours', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
min, max, mean = df['num_commits'].min(), df['num_commits'].max(), df['num_commits'].mean()

print("Number of commit stats")
print(f"min: {min}, max: {max}, mean: {mean}")
```
